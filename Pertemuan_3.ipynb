```python
 #Program menghitung efisiensi termal pembangkit
# Rumus: efisiensi (%) = (energi output / energi input) * 100

energi_input = float(input("Masukkan energi input (MJ): "))
energi_output = float(input("Masukkan energi output (MJ): "))

efisiensi = (energi_output / energi_input) * 100  # tanda kurung memastikan precedence benar

print(f"Efisiensi termal = {efisiensi:.2f}%")
```

```python
#Pertemuan 4
# Input dari pengguna
jumlah_buku = int(input("Masukkan jumlah buku: "))
harga_buku = int(input("Masukkan harga per buku: "))
diskon_persen = int(input("Masukkan diskon (%): "))
uang_budi = int(input("Masukkan uang Budi: "))
total = jumlah_buku * harga_buku
besar_diskon = total * diskon_persen / 100
bayar = total - besar_diskon
kembalian = uang_budi - bayar
iuran = bayar // 4
sisa = bayar % 4
cukup = uang_budi >= bayar

# Output
print("Total belanja:", total)
print("Diskon:", besar_diskon)
print("Jumlah bayar:", bayar)
print("Kembalian:", kembalian)
print("Iuran per orang:", iuran, "Sisa:", sisa)
print("Apakah uang Budi cukup?", cukup)
```

```python
# Program perhitungan PLTS rumah
jumlah_panel = int(input("Masukkan jumlah panel surya: "))
daya_panel = float(input("Masukkan daya per panel (Wp): "))
jam_sinar = float(input("Masukkan rata-rata jam penyinaran efektif per hari: "))
efisiensi = float(input("Masukkan efisiensi sistem (misalnya 0.8): "))
kebutuhan_harian = float(input("Masukkan kebutuhan listrik rumah per hari (kWh): "))
tarif = float(input("Masukkan tarif listrik (Rp/kWh): "))
energi_harian = (jumlah_panel * daya_panel * jam_sinar * efisiensi) / 1000
energi_bulanan = energi_harian * 30
nilai_rupiah = energi_bulanan * tarif
kekurangan = kebutuhan_harian - energi_harian
cukup = energi_harian >= kebutuhan_harian

# Output
print(f"Energi harian: {energi_harian:.2f} kWh")
print(f"Energi bulanan: {energi_bulanan:.2f} kWh")
print(f"Nilai rupiah listrik: Rp {nilai_rupiah:,.2f}")
print(f"Kekurangan energi per hari: {kekurangan:.2f} kWh")
print(f"Apakah PLTS mencukupi kebutuhan rumah? {cukup}")
```

```python
# Program perhitungan generator diesel
daya_generator = float(input("Masukkan daya generator (kW): "))
jam_operasi = float(input("Masukkan jam operasi per hari: "))
konsumsi_per_kWh = float(input("Masukkan konsumsi solar per kWh (liter): "))
harga_solar = float(input("Masukkan harga solar per liter (Rp): "))
kapasitas_tangki = float(input("Masukkan kapasitas tangki (liter): "))
arus = float(input("Masukkan arus kabel (A): "))
hambatan = float(input("Masukkan hambatan kabel (Ohm): "))
energi_harian = daya_generator * jam_operasi
solar_harian = energi_harian * konsumsi_per_kWh
biaya_solar = solar_harian * harga_solar
hari_tangki = kapasitas_tangki // solar_harian
sisa_solar = kapasitas_tangki % solar_harian
rugi_kabel = (arus ** 2) * hambatan
rugi_tinggi = (rugi_kabel > 150) and (arus >= 20)

# Output
print(f"Energi harian: {energi_harian:.2f} kWh")
print(f"Kebutuhan solar per hari: {solar_harian:.2f} liter")
print(f"Biaya solar per hari: Rp {biaya_solar:,.2f}")
print(f"Tangki cukup untuk {hari_tangki:.0f} hari penuh, sisa {sisa_solar:.2f} liter")
print(f"Rugi daya kabel: {rugi_kabel:.2f} W")
print(f"Apakah rugi kabel tinggi? {rugi_tinggi}")
```

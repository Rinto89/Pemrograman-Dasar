```python
#mengambil sebagian dari sebuah daftar
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

print(days[1:4])    # index 1, 2, 3 (index 4 is NOT included)
print(days[:3])     # from the start up to index 2
print(days[4:])     # from index 4 to the end
print(days[-2:])    # the last two elements
print(days[:])      # everything: a copy of the whole list
```

```python
#mengubah menambah dan menghapus elemen
shopping = ["rice", "eggs", "milk"]
print("start        :", shopping)

shopping[1] = "tofu"          # replace the content of box 1
print("after change :", shopping)

shopping.append("tea")        # a new box at the end
print("after append :", shopping)

shopping.insert(0, "salt")    # a new box at index 0; the others shift right
print("after insert :", shopping)

shopping.remove("rice")       # remove the first "rice"
print("after remove :", shopping)

last = shopping.pop()         # remove the LAST element and give it back
print("popped       :", last)
print("after pop    :", shopping)

print("tofu in list?", "tofu" in shopping)
print("rice in list?", "rice" in shopping)
```

```python
# Turbine bearing temperature (°C), one value per hour
temps = [
    61, 60, 60, 59, 59, 60,   # 00:00–05:00
    63, 66, 69, 71, 72, 73,   # 06:00–11:00
    74, 74, 75, 74, 73, 72,   # 12:00–17:00
    70, 68, 66, 64, 63, 62,   # 18:00–23:00
]

print("Values :", len(temps))           # 24 elements: one per hour
print("00:00  :", temps[0], "°C")       # index 0 = the first element
print("14:00  :", temps[14], "°C")      # index = hour, because recording starts at 00:00
print("23:00  :", temps[-1], "°C")      # -1 = the last element
```

```python
#Memotong berdasarkan periode dengan mengiris
temps = [
    61, 60, 60, 59, 59, 60,   # 00:00–05:00
    63, 66, 69, 71, 72, 73,   # 06:00–11:00
    74, 74, 75, 74, 73, 72,   # 12:00–17:00
    70, 68, 66, 64, 63, 62,   # 18:00–23:00
]

night = temps[0:6]       # 00:00–05:00 (index 6 not included)
morning = temps[6:12]    # 06:00–11:00
midday = temps[12:18]    # 12:00–17:00
evening = temps[18:]     # empty stop = up to the last element

print("Morning     :", morning)
print("Midday      :", midday)
print("Every 3 h   :", temps[::3])       # step 3: indices 0, 3, 6, ..., 21
print("Last 6 hours:", temps[-6:])       # the last 6 elements
print("Reversed    :", temps[::-1][:4])  # [::-1] reverses, then [:4] takes the first 4
print("Lengths     :", len(night), len(morning), len(midday), len(evening))
```

```python
#rata-rata, tertinggi, terendah, rentang
temps = [
    61, 60, 60, 59, 59, 60,   # 00:00–05:00
    63, 66, 69, 71, 72, 73,   # 06:00–11:00
    74, 74, 75, 74, 73, 72,   # 12:00–17:00
    70, 68, 66, 64, 63, 62,   # 18:00–23:00
]

average = sum(temps) / len(temps)              # total divided by the number of values
print("Average :", round(average, 1), "°C")    # round(..., 1): one decimal place
print("Highest :", max(temps), "°C")
print("Lowest  :", min(temps), "°C")
print("Range   :", max(temps) - min(temps), "°C")   # highest - lowest

hot_hour = temps.index(max(temps))             # POSITION of the highest = the hour
print("Hottest at", str(hot_hour) + ":00")
print("Midday average:", round(sum(temps[12:18]) / 6, 1), "°C")   # slicing + sum
```

```python
#Mencari: apakah ada di sana, berapa kali, di mana
temps = [
    61, 60, 60, 59, 59, 60,   # 00:00–05:00
    63, 66, 69, 71, 72, 73,   # 06:00–11:00
    74, 74, 75, 74, 73, 72,   # 12:00–17:00
    70, 68, 66, 64, 63, 62,   # 18:00–23:00
]

print(75 in temps)                         # True: it reached 75 °C
print(80 in temps)                         # False: it never reached 80 °C
print("74 °C recorded", temps.count(74), "times")
print("First time at 74 °C:", str(temps.index(74)) + ":00")   # FIRST position
```

```python
#Pengurutan tanpa merusak data asli
temps = [
    61, 60, 60, 59, 59, 60,   # 00:00–05:00
    63, 66, 69, 71, 72, 73,   # 06:00–11:00
    74, 74, 75, 74, 73, 72,   # 12:00–17:00
    70, 68, 66, 64, 63, 62,   # 18:00–23:00
]

ordered = sorted(temps)                # a NEW sorted list
print("Lowest 3 :", ordered[:3])
print("Highest 3:", ordered[-3:])
print("Original :", temps[:3])         # unchanged: still in hourly order

median = (ordered[11] + ordered[12]) / 2   # 24 values: the two in the middle
print("Median   :", median, "°C")

copy = temps.copy()                    # copy first...
copy.sort(reverse=True)                # ...because sort() changes the list itself
print("Largest first:", copy[:5])
```

```python
# Energy sent out per day, 1–30 November 2026 (MWh), 10 days per line
energy = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,   # days 1–10
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,   # days 11–20
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062,   # days 21–30
]
days = list(range(1, 31))     # [1, 2, ..., 30]
run_hours = [24] * 30         # 30 elements, each 24
run_hours[16] = 18            # day 17 (index 16) ran only 18 hours

print(len(energy), len(days), len(run_hours))     # all three are 30
print("Day 17 :", energy[16], "MWh,", run_hours[16], "h")
print("Run hours, days 15–19:", run_hours[14:19])
```

```python
Meringkas bulan dan membaginya per minggu
energy = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,   # days 1–10
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,   # days 11–20
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062,   # days 21–30
]
days = list(range(1, 31))

total = sum(energy)
print("Total   :", total, "MWh")
print("Average :", round(total / len(energy)), "MWh/day")

i_max = energy.index(max(energy))     # position of the highest value
i_min = energy.index(min(energy))
print("Highest : day", days[i_max], "-", energy[i_max], "MWh")   # days[i] = day number
print("Lowest  : day", days[i_min], "-", energy[i_min], "MWh")

cf = total / (330 * 24 * len(energy)) * 100   # actual ÷ maximum × 100
print("Capacity factor:", round(cf, 1), "%")

print("Week 1  :", sum(energy[0:7]), "MWh")     # indices 0–6 = days 1–7
print("Week 3  :", sum(energy[14:21]), "MWh")   # indices 14–20 = days 15–21
print("Sundays :", energy[0::7])                # step 7: days 1, 8, 15, 22, 29
```

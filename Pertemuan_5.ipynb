```python
# Program menghitung skor kualitas daya
# Rumus: skor = faktor_daya * efisiensi

faktor_daya = float(input("Masukkan faktor daya (0.0 - 1.0): "))
efisiensi = float(input("Masukkan efisiensi distribusi (%): "))

# Hitung skor
skor = faktor_daya * efisiensi

# Klasifikasi kualitas daya
if skor >= 85 and faktor_daya >= 0.90:
    print(f"Skor = {skor:.2f}")
    print("Klasifikasi: PRIMA — sistem dalam kondisi optimal")
elif skor >= 70 and faktor_daya >= 0.80:
    print(f"Skor = {skor:.2f}")
    print("Klasifikasi: BAIK — sistem memadai, pantau rutin")
elif skor >= 50 or faktor_daya >= 0.75:
    print(f"Skor = {skor:.2f}")
    print("Klasifikasi: CUKUP — ada ruang perbaikan")
else:
    print(f"Skor = {skor:.2f}")
    print("Klasifikasi: BURUK — perlu perbaikan segera")
```

```python
# Program menentukan mode operasi PLTS
# Rumus daya panel: daya_panel = iradiasi * 0.18 * 10

iradiasi = float(input("Masukkan iradiasi matahari (W/m²): "))
soc = float(input("Masukkan SoC baterai (%): "))

# Hitung daya panel
daya_panel = iradiasi * 0.18 * 10

print(f"Daya panel = {daya_panel:.2f} W")

# Tentukan mode operasi
if iradiasi > 600 and soc < 80:
    print("Mode Operasi: CHARGING — iradiasi tinggi, baterai belum penuh, prioritaskan pengisian")
elif iradiasi > 600 and soc >= 80:
    print("Mode Operasi: GRID-EXPORT — baterai sudah penuh, ekspor kelebihan daya ke jaringan")
elif iradiasi > 200 and soc >= 20:
    print("Mode Operasi: NORMAL — iradiasi dan baterai cukup, operasi mandiri")
elif iradiasi <= 200 and soc >= 50:
    print("Mode Operasi: BATTERY-ONLY — iradiasi rendah, andalkan baterai")
else:
    print("Mode Operasi: EMERGENCY — iradiasi rendah dan baterai hampir habis, kurangi beban")
```

```python
#Soal no 8

# Program simulasi pengosongan tangki bahan bakar

# Volume awal tangki 100 liter
volume = 100

# Perulangan berjalan selama volume masih lebih dari 0
while volume > 0:

    # Mengurangi volume sebanyak 20 liter setiap putaran
    volume = volume - 20

    # Menampilkan volume setelah dikurangi
    print(f"Volume: {volume} liter")

    # Jika volume tepat 20 liter, tampilkan peringatan
    if volume == 20:
        print("WARNING: Hampir kosong!")

# Setelah perulangan selesai (volume = 0)
print("TANGKI KOSONG - Pengosongan berhenti.")
```

```python
#soal no 10

# Data suhu bearing dalam bentuk teks
data = ["62.5", "58.0", "81.3", "N/A", "66.0"]

# Menghitung jumlah data yang berhasil diperiksa
jumlah_valid = 0

# Memeriksa setiap data beserta nomor titik pengukurannya
for i, nilai in enumerate(data, start=1):

    try:
        # Mengubah data teks menjadi tipe float
        suhu = float(nilai)

        # Menambah jumlah data valid jika konversi berhasil
        jumlah_valid += 1

        # Memeriksa apakah suhu melebihi 70°C
        if suhu > 70:
            print(f"Warning: Titik {i} suhu bearing tinggi!")

    except ValueError:
        # Menampilkan pesan jika data tidak valid
        print(f"Titik {i}: data tidak valid - pemeriksaan dihentikan")

        # Menghentikan perulangan saat menemukan data tidak valid
        break

# Menampilkan jumlah data valid yang berhasil diperiksa
print(f"Jumlah data valid: {jumlah_valid}")
```

```python
#Soal chalenge

# Program pemantauan konsumsi listrik harian rumah

# Langkah 1: Menyimpan data konsumsi listrik harian (kWh) dalam list
data_kwh = ["12.5", "15.2", "28.7", "32.1", "error", "18.4"]

# Langkah 2: Membuat variabel untuk menghitung total konsumsi
total_kwh = 0

# Langkah 3: Membuat variabel untuk menghitung jumlah data valid
jumlah_data = 0

# Langkah 4: Membaca setiap data konsumsi listrik
for hari, nilai in enumerate(data_kwh, start=1):

    try:
        # Mengubah data string menjadi float
        konsumsi = float(nilai)

        # Menambahkan konsumsi ke total
        total_kwh += konsumsi

        # Menambah jumlah data valid
        jumlah_data += 1

        # Menentukan status konsumsi listrik
        if konsumsi > 30:
            print(f"Hari {hari}: BOROS ({konsumsi} kWh)")
        elif konsumsi >= 20:
            print(f"Hari {hari}: SEDANG ({konsumsi} kWh)")
        else:
            print(f"Hari {hari}: HEMAT ({konsumsi} kWh)")

    except ValueError:
        # Jika data tidak dapat diubah menjadi angka
        print(f"Hari {hari}: Data tidak valid ({nilai})")

# Langkah 5: Menghitung rata-rata konsumsi listrik
if jumlah_data > 0:
    rata_rata = total_kwh / jumlah_data

    print("\n=== HASIL PEMANTAUAN ===")
    print(f"Total konsumsi   : {total_kwh:.1f} kWh")
    print(f"Jumlah data valid: {jumlah_data}")
    print(f"Rata-rata        : {rata_rata:.2f} kWh")
else:
    print("Tidak ada data valid.")
```

```python
#fungsi dari setiap Code yang ada pada soal chalenge

# Program pemantauan konsumsi listrik harian rumah

# Menyimpan data konsumsi listrik harian dalam bentuk list
data_kwh = ["12.5", "15.2", "28.7", "32.1", "error", "18.4"]

# Variabel untuk menyimpan total konsumsi listrik
total_kwh = 0

# Variabel untuk menghitung jumlah data yang valid
jumlah_data = 0

# Melakukan perulangan untuk membaca setiap data pada list
# enumerate digunakan untuk memberikan nomor hari dimulai dari 1
for hari, nilai in enumerate(data_kwh, start=1):

    try:
        # Mengubah data string menjadi tipe float
        konsumsi = float(nilai)

        # Menambahkan nilai konsumsi ke total konsumsi
        total_kwh += konsumsi

        # Menambah jumlah data valid
        jumlah_data += 1

        # Memeriksa apakah konsumsi listrik lebih dari 30 kWh
        if konsumsi > 30:

            # Menampilkan status BOROS
            print(f"Hari {hari}: BOROS ({konsumsi} kWh)")

        # Memeriksa apakah konsumsi listrik antara 20 sampai 30 kWh
        elif konsumsi >= 20:

            # Menampilkan status SEDANG
            print(f"Hari {hari}: SEDANG ({konsumsi} kWh)")

        # Jika konsumsi kurang dari 20 kWh
        else:

            # Menampilkan status HEMAT
            print(f"Hari {hari}: HEMAT ({konsumsi} kWh)")

    # Menangani kesalahan jika data tidak dapat diubah menjadi angka
    except ValueError:

        # Menampilkan pesan bahwa data tidak valid
        print(f"Hari {hari}: Data tidak valid ({nilai})")

# Memeriksa apakah terdapat data valid yang berhasil diproses
if jumlah_data > 0:

    # Menghitung rata-rata konsumsi listrik
    rata_rata = total_kwh / jumlah_data

    # Menampilkan hasil akhir pemantauan
    print("\n=== HASIL PEMANTAUAN ===")
    print(f"Total konsumsi   : {total_kwh:.1f} kWh")
    print(f"Jumlah data valid: {jumlah_data}")
    print(f"Rata-rata        : {rata_rata:.2f} kWh")

# Jika tidak ada data valid
else:

    # Menampilkan pesan bahwa tidak ada data yang dapat dihitung
    print("Tidak ada data valid.")
```

```python
#cara kerja program untuk soal chalenge

# Program pemantauan konsumsi listrik harian rumah

# Langkah 1: Menyimpan data konsumsi listrik harian dalam bentuk list
data_kwh = ["12.5", "15.2", "28.7", "32.1", "error", "18.4"]

# Langkah 2: Membuat variabel untuk menyimpan total konsumsi listrik
total_kwh = 0

# Langkah 3: Membuat variabel untuk menghitung jumlah data valid
jumlah_data = 0

# Langkah 4: Melakukan perulangan untuk membaca setiap data dalam list
# enumerate() digunakan untuk memberikan nomor hari mulai dari 1
for hari, nilai in enumerate(data_kwh, start=1):

    try:
        # Langkah 5: Mengubah data string menjadi tipe float
        konsumsi = float(nilai)

        # Langkah 6: Menambahkan nilai konsumsi ke total konsumsi listrik
        total_kwh += konsumsi

        # Langkah 7: Menambah jumlah data yang valid
        jumlah_data += 1

        # Langkah 8: Memeriksa kategori konsumsi listrik menggunakan if-elif-else
        if konsumsi > 30:
            # Jika konsumsi lebih dari 30 kWh maka status BOROS
            print(f"Hari {hari}: BOROS ({konsumsi} kWh)")

        elif konsumsi >= 20:
            # Jika konsumsi antara 20–30 kWh maka status SEDANG
            print(f"Hari {hari}: SEDANG ({konsumsi} kWh)")

        else:
            # Jika konsumsi kurang dari 20 kWh maka status HEMAT
            print(f"Hari {hari}: HEMAT ({konsumsi} kWh)")

    except ValueError:
        # Langkah 9: Menangani kesalahan jika data tidak dapat diubah menjadi float
        print(f"Hari {hari}: Data tidak valid ({nilai})")

# Langkah 10: Setelah perulangan selesai, periksa apakah ada data valid
if jumlah_data > 0:

    # Langkah 11: Menghitung rata-rata konsumsi listrik
    rata_rata = total_kwh / jumlah_data

    # Langkah 12: Menampilkan hasil akhir pemantauan
    print("\n=== HASIL PEMANTAUAN ===")
    print(f"Total konsumsi   : {total_kwh:.1f} kWh")
    print(f"Jumlah data valid: {jumlah_data}")
    print(f"Rata-rata        : {rata_rata:.2f} kWh")

else:
    # Jika tidak ada data valid yang berhasil diproses
    print("Tidak ada data valid.")

# Langkah 13: Program selesai
```
